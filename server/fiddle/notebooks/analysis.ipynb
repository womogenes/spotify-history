```python
import pandas as pd
import matplotlib.pyplot as plt

from fiddle.scripts.streams import get_all_streams
```

```python
all_streams = get_all_streams()
```

```python
df = pd.DataFrame.from_dict(all_streams)
df.head()

# Retype the ts field
df.ts = pd.to_datetime(df.ts)
```

```python
fig, ax = plt.subplots(1, 1, figsize=(18, 4))
plt.hist(df.ts, weights=df.ms_played, bins=200);
plt.xlabel("Date")
plt.ylabel("ms music listened to")
plt.title("My Spotify listening history");
```
